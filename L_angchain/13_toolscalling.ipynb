# custom tools

```python
from langchain_core.tools import tool

@tool # decorator
def add(a: int,b:int) ->int: # type hints
    """add two numbers"""
    return a+b   

print(add.invoke({"a":3, "b": 7}))  
print("....................")
print(add.name,"..." ,add.description,"....", add.args)  
```

### Using StructuredTool(custom)

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field

class addinput(BaseModel):
    a:int = Field(required=True, description="1st number")
    b:int = Field(required=True, description="2nd number")

def add_fun(a:int, b:int) -> int:
    return a+b

add_tool = StructuredTool.from_function(
    func = add_fun,
    name='add',
    description='add two numbers',
    args_schema= addinput
)    

res = add_tool.invoke({"a":456,"b":3})
print(res)
print(add_tool.name)
print(add_tool.description)
print(add_tool.args)
```

### Using BaseTool Class

```python
from langchain.tools import BaseTool
from typing import Type

# arg schema using pydantic

class MultiplyInput(BaseModel):
    a: int = Field(required=True, description="The first number to add")
    b: int = Field(required=True, description="The second number to add")

class MultiplyTool(BaseTool):
    name: str = "multiply"
    description: str = "Multiply two numbers"

    args_schema: Type[BaseModel] = MultiplyInput

    def _run(self, a: int, b: int) -> int:
        return a * b   

multiply_tool = MultiplyTool()


result = multiply_tool.invoke({'a':3, 'b':3})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)

print(multiply_tool.args)         
```

#### toolkit --- having multiple tools

```python
from langchain_core.tools import tool

# Custom tools
@tool
def add(a: int, b: int) -> int:
    """Add two numbers"""
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a * b


class MathToolkit:
    def get_tools(self):
        return [add, multiply]


toolkit = MathToolkit()
tools = toolkit.get_tools()

for tool in tools:
    print(tool.name, "=>", tool.description)

```
